```python
def add(p,q,dis=0.05):
  total=p*q
  tax=total*dis
  totoal_all=total+tax
  return total,tax,totoal_all

p=int(input("請輸入單價:"))
q=int(input("請輸入數量:"))
x,y,z=add(p,q)
print(f"小計:{x}\n稅金:{y}\n總金額:{z}")
```
